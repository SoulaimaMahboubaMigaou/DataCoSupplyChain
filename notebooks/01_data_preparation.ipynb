:::::::::::::::::::: ETAPE 2: DATA CLEANING :::::::::::::::::

```python
df["shipping date (DateOrders)"] = pd.to_datetime(df["shipping date (DateOrders)"])
```

```python
numeric_cols = df.select_dtypes(include=["number"]).columns

df[numeric_cols] = df[numeric_cols].apply(
    pd.to_numeric, errors="coerce"
)
```

```python
df.dtypes
```

SUPPRIMER LES COLONNES INUTILISABLES :

```python
df = df.drop(columns=["Product Description", "Order Zipcode"])
```

TRAITER LES VALEURS MANQUANTES RESTANTES :

```python
# 'Customer Lname' (8 manquants) : on remplace par "Unknown"
df["Customer Lname"] = df["Customer Lname"].fillna("Unknown")
```

```python
# 'Customer Zipcode' (3 manquants) : on remplace par la médiane
df["Customer Zipcode"] = df["Customer Zipcode"].fillna(df["Customer Zipcode"].median())
```

SUPPRIMER LES DOUBLONS :

```python
before = df.shape[0]
df = df.drop_duplicates()
print(f"\nDoublons supprimés : {before - df.shape[0]}")
```

NETTOYER LES NOMS DE COLONNES (plus faciles à manipuler) :

```python
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)
```

CONVERTIR LES DATES EN VRAI FORMAT DATETIME :

```python
df["order_date_dateorders"] = pd.to_datetime(df["order_date_dateorders"])
```

```python
df["shipping_date_dateorders"] = pd.to_datetime(df["shipping_date_dateorders"])
```

 SUPPRIMER LES COLONNES SENSIBLES / INUTILES (optionnel) :

```python
df = df.drop(columns=["customer_email", "customer_password"])
```

VÉRIFICATION FINALE :

```python
print("\nShape finale :", df.shape)
print("\nValeurs manquantes restantes :")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nTypes de données :")
print(df.dtypes)
```

:::::::::::::::::: ETAPE 3 :Feature Engineering :::::::::::::

```python
df.columns.tolist()
```

```python
df[["order_date_dateorders", "shipping_date_dateorders"]].head(10)
```

A PARTIR DE order_date_dateorders (Annees) :

```python
df["order_year"] = df["order_date_dateorders"].dt.year
```

A PARTIR DE order_date_dateorders (Mois) :

```python
df["order_month"] = df["order_date_dateorders"].dt.month
```

```python
df["order_month_name"] = df["order_date_dateorders"].dt.month_name()
```

```python
df["order_day_of_week"] = df["order_date_dateorders"].dt.day_name()
```

```python
df["order_day"] = df["order_date_dateorders"].dt.day
```

```python
df["order_quarter"] = df["order_date_dateorders"].dt.quarter
```

Créer delivery_delay,
la différence entre Days for shipping (real) et Days for shipment (scheduled) :

```python
df.columns.tolist()
```

```python
df["delivery_delay"] = (
    df["days_for_shipping_real"]
    - df["days_for_shipment_scheduled"]
)
```

Créer delivery_status :(on transforme le délai en catégorie)

```python
df["delivery_status"] = np.where(
    df["delivery_delay"] > 0,
    "Late",
    "On Time"
)
```

```python
df["delivery_status"].value_counts()
```

Créer une catégorie de retard :

```python
df["delay_category"] = pd.cut(
    df["delivery_delay"],
    bins=[-float("inf"), 0, 2, 5, float("inf")],
    labels=[
        "On Time",
        "Minor Delay",
        "Moderate Delay",
        "Severe Delay"
    ]
)
```

```python
df["delay_category"].value_counts()  #check the distribution of delay categories
```

Créer des catégories de quantité:

```python
df["quantity_category"] = pd.cut(
    df["order_item_quantity"],
    bins=[0, 1, 3, 5, float("inf")],
    labels=[
        "Single Item",
        "Small Order",
        "Medium Order",
        "Bulk Order"
    ]
)
```

Créer une catégorie de valeur de commande:

```python
df["sales_category"] = pd.qcut(
    df["sales"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ],
    duplicates="drop"
)
```

Créer une variable de profit:

```python
df["profit_category"] = pd.cut(
    df["order_item_profit_ratio"],
    bins=[
        -float("inf"),
        0,
        0.1,
        0.2,
        float("inf")
    ],
    labels=[
        "Loss",
        "Low Profit",
        "Medium Profit",
        "High Profit"
    ]
)
```

Créer une variable de risque:

```python
df["delivery_risk"] = df["late_delivery_risk"].map({
    0: "Low Risk",
    1: "High Risk"
})
```
